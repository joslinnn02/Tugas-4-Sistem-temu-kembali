```python
#import library
import re
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
```

```python
#Dataset
dokumen = [
    "Teknologi informasi membantu mahasiswa mencari materi kuliah dengan lebih cepat dan mudah.",
    "Sistem komputer digunakan untuk mengolah data, menyimpan informasi, dan menjalankan berbagai aplikasi.",
    "Jaringan komputer menghubungkan perangkat agar dapat bertukar data dan berbagi sumber daya.",
    "Kecerdasan buatan dapat membantu proses pembelajaran dengan memberikan informasi dan rekomendasi kepada mahasiswa.",
    "Sistem temu kembali informasi digunakan untuk menemukan dokumen yang sesuai dengan kebutuhan pengguna."
]

for i, doc in enumerate(dokumen, 1):
    print(f"Dokumen {i}: {doc}")
```

```python
#membuat stopword dan stemmer
stopword_factory = StopWordRemoverFactory()
stopwords = set(stopword_factory.get_stop_words())

stemmer_factory = StemmerFactory()
stemmer = stemmer_factory.create_stemmer()

print("Stopwords dan stemmer berhasil dibuat.")
```

```python
#fungsi untuk melakukan preprocessing pada teks
def preprocess_text(text):
    # 1. Case folding
    text = text.lower()

    # 2. Cleaning: hapus angka, tanda baca, dan karakter khusus
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()

    # 3. Tokenisasi
    tokens = text.split()

    # 4. Stopwords removal
    tokens = [token for token in tokens if token not in stopwords]

    # 5. Stemming
    tokens = [stemmer.stem(token) for token in tokens]

    return tokens
```

```python
# Preprocessing dokumen
hasil_preprocessing = []

for i, doc in enumerate(dokumen, 1):
    # Token sebelum preprocessing
    tokens_sebelum = re.findall(r'\b[a-zA-Z]+\b', doc.lower())

    # Token setelah preprocessing
    tokens_sesudah = preprocess_text(doc)

    hasil_preprocessing.append({
        "Dokumen": i,
        "Sebelum": doc,
        "Sesudah": " ".join(tokens_sesudah),
        "Token Sebelum": len(tokens_sebelum),
        "Token Sesudah": len(tokens_sesudah)
    })

for hasil in hasil_preprocessing:
    print(f"\nDokumen {hasil['Dokumen']}")
    print("Sebelum :", hasil["Sebelum"])
    print("Sesudah :", hasil["Sesudah"])
```

```python
# Menampilkan hasil preprocessing untuk 2 dokumen pertama
for hasil in hasil_preprocessing[:2]:
    print("=" * 70)
    print(f"DOKUMEN {hasil['Dokumen']}")
    print("Sebelum preprocessing:")
    print(hasil["Sebelum"])
    print("\nSesudah preprocessing:")
    print(hasil["Sesudah"])
    print("=" * 70)
```

```python
# Menampilkan tabel perbandingan jumlah token sebelum dan sesudah preprocessing
print(f"{'Dokumen':<10}{'Token Sebelum':<18}{'Token Sesudah':<18}{'Pengurangan':<15}")
print("-" * 61)

for hasil in hasil_preprocessing:
    sebelum = hasil["Token Sebelum"]
    sesudah = hasil["Token Sesudah"]

    pengurangan = ((sebelum - sesudah) / sebelum) * 100 if sebelum > 0 else 0

    print(
        f"{hasil['Dokumen']:<10}"
        f"{sebelum:<18}"
        f"{sesudah:<18}"
        f"{pengurangan:.2f}%"
    )
```
